```python
def reverse(n):
  rev=0
  while(n>0):
    rem=n%10
    rev=rev*10+rem
    n//=10
  return rev
num=int(input())
if(num==reverse(num)):
  print("Palindrome")
else:
  print("Not Palindrome")
```

```python
def reverse(s):
  return s[::-1]
n=input()
print(reverse(n))
```

```python
def reverse(s):
  rev=""
  for i in range(len(n)-1,-1,-1):
    rev=rev+s[i]
  return rev
n=input()
print(reverse(n))
```

```python
#Example on variable length arguments
def add(*num):
  print(num)
  print(sum(num))
add(1,2,3,4,5)
```

```python
#Function that returns multiple values
def multiple_values(*num):
  c=len(num)
  s=sum(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_values(1,2,3,4,5)
print(f"Sum is {res_sum},No.of digits {res_count},AVG of numbers {res_avg}")
```
